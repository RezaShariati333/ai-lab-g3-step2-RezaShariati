```python
import pandas as pd
```

```python
# خواندن اطلاعات فیلم‌ها و امتیازهای کاربران
movies = pd.read_csv("movies.csv")
ratings = pd.read_csv("ratings.csv")

print("فایل‌ها با موفقیت خوانده شدند.")
```

```python
# نمایش نام ستون‌ها
print("ستون‌های فیلم‌ها:")
print(movies.columns.tolist())

print("\nستون‌های امتیازها:")
print(ratings.columns.tolist())

# بررسی نوع داده‌های هر ستون
print("\nنوع داده‌های فیلم‌ها:")
print(movies.dtypes)

print("\nنوع داده‌های امتیازها:")
print(ratings.dtypes)

# نمایش نمونه‌ای از اطلاعات
display(movies.head())
display(ratings.head())

# بررسی ابعاد جدول‌ها
print("ابعاد فیلم‌ها:", movies.shape)
print("ابعاد امتیازها:", ratings.shape)

# شمارش فیلم‌ها، کاربران و امتیازها
print("تعداد فیلم‌های یکتا:", movies["movieId"].nunique())
print("تعداد کاربران یکتا:", ratings["userId"].nunique())
print("تعداد کل امتیازها:", len(ratings))
```

```python
# جدا کردن ژانرهای هر فیلم
genres = movies["genres"].str.split("|").explode()

# شمارش تعداد فیلم‌ها در هر ژانر
genre_counts = genres.value_counts()

print("تعداد فیلم‌ها در هر ژانر:")
display(genre_counts.to_frame(name="movie_count"))
```

```python
# نمایش خلاصه نتایج بررسی داده‌ها
print("===== گزارش بررسی اولیه داده‌ها =====")

print("\nتعداد فیلم‌ها:", len(movies))
print("تعداد کاربران:", ratings["userId"].nunique())
print("تعداد کل امتیازها:", len(ratings))

print("\nبررسی اولیه داده‌ها به پایان رسید.")
```
